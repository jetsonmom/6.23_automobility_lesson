```python
# 복잡한 코드 이 코드는 센서 기반 자율주행 자동차 시뮬레이션 코드
# 1. 센서 초기화 함수
def init_sensors():
    """센서 기본값 설정"""
    sensors = {
        "앞": 15.0,      # 기본값
        "왼쪽": 15.0,
        "오른쪽": 15.0
    }
    print("10 - 센서 초기화 완료", sensors)
    return sensors

# 2. 앞쪽 스캔 함수
def scan_front(x, y, obstacles, sensors):
    """앞쪽 장애물 스캔"""
    print("16- 앞쪽 스캔 시작")
    if (x, y+1) in obstacles:
        print("17", x, y, "앞쪽 1칸 장애물 발견")
        sensors["앞"] = 1.0
    elif (x, y+2) in obstacles:
        sensors["앞"] = 2.0
        print("22", x, y, "앞쪽 2칸 장애물 발견")
    elif (x, y+3) in obstacles:
        sensors["앞"] = 3.0
        print("25", x, y, "앞쪽 3칸 장애물 발견")
    else:
        print("27", x, y, "앞쪽 장애물 없음")
    print("28 - 앞쪽 스캔 완료")

# 3. 왼쪽 스캔 함수
def scan_left(x, y, obstacles, sensors):
    """왼쪽 장애물 스캔"""
    print("33- 왼쪽 스캔 시작")
    if (x-1, y) in obstacles:
        sensors["왼쪽"] = 1.0
        print("36", x, y, "왼쪽 1칸 장애물 발견")
    elif (x-2, y) in obstacles:
        sensors["왼쪽"] = 2.0
        print("39", x, y, "왼쪽 2칸 장애물 발견")
    elif (x-3, y) in obstacles:
        sensors["왼쪽"] = 3.0
        print("42", x, y, "왼쪽 3칸 장애물 발견")
    else:
        print("44", x, y, "왼쪽 장애물 없음")
    print("45 - 왼쪽 스캔 완료")

# 4. 오른쪽 스캔 함수
def scan_right(x, y, obstacles, sensors):
    """오른쪽 장애물 스캔"""
    print("50 - 오른쪽 스캔 시작")
    if (x+1, y) in obstacles:
        sensors["오른쪽"] = 1.0
        print("53", x, y, "오른쪽 1칸 장애물 발견")
    elif (x+2, y) in obstacles:
        sensors["오른쪽"] = 2.0
        print("56", x, y, "오른쪽 2칸 장애물 발견")
    elif (x+3, y) in obstacles:
        sensors["오른쪽"] = 3.0
        print("59", x, y, "오른쪽 3칸 장애물 발견")
    else:
        print("61", x, y, "오른쪽 장애물 없음")
    print("62 - 오른쪽 스캔 완료")

# 5. 센서 데이터 업데이트 함수 (디버그 print 포함)
def update_sensor_data(position, obstacles):
    """현재 위치에서 센서 데이터 업데이트"""
    print("67 - update_sensor_data 함수 시작")
    x, y = position["x"], position["y"]
    print("69- 위치 추출 완료")
    print("70", x, y)

    # 센서 초기화
    sensors = init_sensors()

    # 각 방향별 스캔 실행
    scan_front(x, y, obstacles, sensors)
    scan_left(x, y, obstacles, sensors)
    scan_right(x, y, obstacles, sensors)

    print("80 - 센서 업데이트 완료", sensors)
    return sensors

# 6. 장애물 감지 함수
def detect_obstacles(sensors, safe_distance):
    """장애물을 감지하는 함수 (원본 그대로)"""
    print("86 - detect_obstacles 함수 시작")
    print("87", sensors, safe_distance)
    obstacles = []

    for direction, distance in sensors.items():
        print("91", direction, distance, "체크 중")
        if distance < safe_distance:
            obstacles.append(direction)
            print("94", direction, "위험! 장애물 목록에 추가")
        else:
            print("96", direction, "안전")

    print("98 - 장애물 감지 완료", obstacles)
    return obstacles

# 7. 행동 결정 함수
def decide_action(obstacles):
    """장애물 정보를 바탕으로 행동을 결정하는 함수 (원본 그대로)"""
    print("104 - decide_action 함수 시작")
    print("105", obstacles, "입력된 장애물 목록")

    if len(obstacles) == 0: # len()은 Python의 내장 함수로, 객체의 길이(length), obstacles 리스트(또는 다른 컬렉션)가 비어있는지 확인하는 조건문
        action = "직진"
        print("109", "장애물 없음 -> 직진")
    elif "앞" in obstacles:
        print("111", "앞에 장애물 있음, 좌우 확인")
        if "왼쪽" not in obstacles:
            action = "좌회전"
            print("114", "왼쪽 안전 -> 좌회전")
        elif "오른쪽" not in obstacles:
            action = "우회전"
            print("117", "오른쪽 안전 -> 우회전")
        else:
            action = "정지"
            print("120", "모든 방향 위험 -> 정지")
    else:
        action = "주의해서 직진"
        print("123", "앞은 안전, 옆만 위험 -> 주의해서 직진")

    print("125 - 행동 결정 완료", action)
    return action

# 8. 간단한 자율주행 자동차 클래스 (디버그 포함)
class SimpleCar:
    def __init__(self, name, x, y):
        """자동차 생성"""
        print("132 - SimpleCar __init__ 시작")
        self.name = name
        self.position = {"x": x, "y": y}
        self.safe_distance = 10.0
        print("136", name, x, y, "자동차 생성 완료")

    def move_forward(self):
        """전진"""
        print("140 - move_forward 시작")
        self.position["y"] += 1
        print("142", self.position["x"], self.position["y"], "전진 완료")
        print(f"🚗 {self.name}이(가) 전진합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def turn_left(self):
        """좌회전"""
        print("147 - turn_left 시작")
        self.position["x"] -= 1
        print("149", self.position["x"], self.position["y"], "좌회전 완료")
        print(f"↖️ {self.name}이(가) 좌회전합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def turn_right(self):
        """우회전"""
        print("154 - turn_right 시작")
        self.position["x"] += 1
        print("156", self.position["x"], self.position["y"], "우회전 완료")
        print(f"↗️ {self.name}이(가) 우회전합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def stop(self):
        """정지"""
        print("161 - stop 시작")
        print("162", self.position["x"], self.position["y"], "정지")
        print(f"🛑 {self.name}이(가) 정지합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def execute_action(self, action):
        """행동 실행"""
        print("167 - execute_action 시작")
        print("168", action, "실행할 행동")

        if action == "직진" or action == "주의해서 직진":
            print("171", "직진 관련 행동 -> move_forward 호출")
            self.move_forward()
        elif action == "좌회전":
            print("174", "좌회전 -> turn_left 호출")
            self.turn_left()
        elif action == "우회전":
            print("177", "우회전 -> turn_right 호출")
            self.turn_right()
        elif action == "정지":
            print("180", "정지 -> stop 호출")
            self.stop()

        print("183 - execute_action 완료")

    def run_simulation(self, obstacles):
        """자율주행 시뮬레이션"""
        print("187 - run_simulation 시작")
        steps = 0
        max_steps = 10

        print("🚗 자율주행 장애물 감지 시스템")
        print("-" * 30)
        print(f"🚧 장애물 위치: {list(obstacles)}")
        print("194", obstacles, "시뮬레이션 시작")

        while steps < max_steps:
            print(f"\n--- 스텝 {steps + 1} ---")
            print("198", steps + 1, "스텝 시작")
            print(f"📍 현재 위치: ({self.position['x']}, {self.position['y']})")
            print("200", self.position["x"], self.position["y"], "현재 위치")

            # 센서 데이터 업데이트
            print("203 - 센서 업데이트 호출")
            sensors = update_sensor_data(self.position, obstacles)
            print(f"📡 센서 데이터: {sensors}")
            print("20", sensors, "센서 데이터 받음")

            # 장애물 감지 (원본 함수 그대로 사용)
            print("2 - 장애물 감지 호출")
            detected_obstacles = detect_obstacles(sensors, self.safe_distance)
            print(f"🚨 감지된 장애물: {detected_obstacles}")
            print("212", detected_obstacles, "감지된 장애물")

            # 행동 결정 (원본 함수 그대로 사용)
            print("215 - 행동 결정 호출")
            action = decide_action(detected_obstacles)
            print(f"🎯 결정된 행동: {action}")
            print("218", action, "결정된 행동")

            # 행동 실행
            print("221 - 행동 실행 호출")
            self.execute_action(action)

            # 정지했으면 종료
            if action == "정지":
                print("226", "정지 조건 만족, 시뮬레이션 종료")
                print("⚠️ 모든 방향 막혀서 시뮬레이션 종료!")
                break

            steps += 1
            print("231", steps, "스텝 완료")

        print(f"\n🏁 시뮬레이션 완료!")
        print("234 - run_simulation 완료")

# 9. 시뮬레이션 실행
if __name__ == "__main__":
    print("238- 메인 실행 시작")

    # 자동차 생성
    print("241- 자동차 생성")
    car = SimpleCar("현대차", 3, 0)

    # 장애물 설정
    obstacles = {(3, 2), (3, 3), (2, 4), (4, 5)}
    print("247", obstacles, "장애물 설정 완료")

    # 시뮬레이션 실행
    car.run_simulation(obstacles)

```

1. 📊 기본 플롯 설정

```python
import matplotlib.pyplot as plt

def setup_basic_plot(step_num=1, action="forward"):
    """기본 플롯 설정만 하는 함수"""
    plt.figure(figsize=(10, 8))
    ax = plt.gca()

    # 축 범위 설정
    ax.set_xlim(-1, 8)
    ax.set_ylim(-1, 12)

    # 플롯 스타일 설정
    ax.set_aspect('equal')  # 정사각형 비율
    ax.grid(True, alpha=0.3)  # 격자 표시

    # 라벨과 제목
    ax.set_xlabel('X Position', fontsize=12)
    ax.set_ylabel('Y Position', fontsize=12)
    ax.set_title(f"Step {step_num} - Action: {action}",
                fontsize=14, fontweight='bold')

    plt.tight_layout()
    plt.show()
    return ax

# 사용 예시
if __name__ == "__main__":
    # 기본 플롯만 보기
    setup_basic_plot(step_num=5, action="turn_left")

    # 다른 설정으로 보기
    setup_basic_plot(step_num=10, action="forward")
```

```python
import matplotlib.pyplot as plt

def draw_goal_line(obstacles):
    """목표선만 그리는 함수"""
    plt.figure(figsize=(8, 6))
    ax = plt.gca()

    # 기본 설정
    ax.set_xlim(-1, 8)
    ax.set_ylim(-1, 12)
    ax.grid(True, alpha=0.3)
    ax.set_title("Goal Line", fontsize=14, fontweight='bold')

    # 목표선 Y 좌표 계산 (가장 높은 장애물 + 3)
    max_obstacle_y = max(pos[1] for pos in obstacles) if obstacles else 0
    goal_y = max_obstacle_y + 3

    # 목표선 그리기 (초록색 수평선)
    ax.axhline(y=goal_y, color='lime', linestyle='-',
              linewidth=3, alpha=0.7, label='Goal Line')

    # 목표선 텍스트
    ax.text(7, goal_y + 0.2, f'GOAL (y={goal_y})',
           ha='right', va='bottom', fontweight='bold',
           color='lime', fontsize=12)


    ax.legend()
    plt.tight_layout()
    plt.show()

    print(f"목표선 Y 좌표: {goal_y}")
    return goal_y

# 실핼 부분
if __name__ == "__main__":
    # 장애물 정의
    obstacles = {(3, 2), (3, 3), (2, 4), (4, 5)}  # 최대 Y = 5
    goal_y = draw_goal_line(obstacles)  # 결과: 5 + 3 = 8



```

3. 🔴 장애물들 48-50, 37-38명령을 활성화 시켜서 테스트도 해봅니다.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def draw_obstacles(obstacles):
    """Draw obstacles visualization - English code with clean display"""
    plt.figure(figsize=(8, 6))
    ax = plt.gca()

    # Basic plot setup
    ax.set_xlim(-1, 8)
    ax.set_ylim(-1, 12)
    ax.grid(True, alpha=0.3)
    ax.set_title("Obstacles", fontsize=14, fontweight='bold')
    ax.set_aspect('equal')
    ax.set_xlabel('X Position', fontsize=12)
    ax.set_ylabel('Y Position', fontsize=12)

    # Convert set to list for consistent ordering
    if isinstance(obstacles, set):
        obstacles_list = list(obstacles)
    else:
        obstacles_list = list(obstacles)

    # Draw each obstacle
    for i, (obs_x, obs_y) in enumerate(obstacles_list):
        # Create red rectangle (0.8 x 0.8 size)
        rect = Rectangle((obs_x-0.4, obs_y-0.4), 0.8, 0.8,
                        color='red', alpha=0.7,
                        label='Obstacle' if i == 0 else "")
        ax.add_patch(rect)

        # Add obstacle number (white text)
        ax.text(obs_x, obs_y, str(i+1), ha='center', va='center',
               color='white', fontweight='bold', fontsize=12)

        # Add coordinate label next to obstacle
   #     ax.text(obs_x + 0.6, obs_y + 0.6, f'({obs_x},{obs_y})',
   #            fontsize=9, fontweight='bold',
   #            bbox=dict(boxstyle='round,pad=0.2', facecolor='yellow', alpha=0.7))

    # Create information box
    data_type = "set" if isinstance(obstacles, set) else "list"
    info_text = f"Total: {len(obstacles)} obstacles ({data_type})\n"
    for i, (x, y) in enumerate(obstacles_list):
        info_text += f"Obstacle {i+1}: ({x}, {y})\n"

    # Display info box in top-left corner
  #  ax.text(0.02, 0.98, info_text, transform=ax.transAxes,
  #         verticalalignment='top',
  #         bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.8))

    # Add legend if obstacles exist
    if obstacles:
        ax.legend(loc='upper right')

    plt.tight_layout()
    plt.show()

    # Print information to console
    print(f"Number of obstacles: {len(obstacles)} ({data_type})")
    for i, pos in enumerate(obstacles_list):
        print(f"Obstacle {i+1}: {pos}")

# Example usage
if __name__ == "__main__":
    # Original code example using set
    obstacles = {(3, 2), (3, 3), (2, 4), (4, 5)}
    draw_obstacles(obstacles)
```

4. 이동경로 그리기

```python
import matplotlib.pyplot as plt

def draw_path(path):
    """이동 경로만 그리는 함수"""
    plt.figure(figsize=(8, 6))
    ax = plt.gca()

    # 기본 설정
    ax.set_xlim(-1, 8)
    ax.set_ylim(-1, 12)
    ax.grid(True, alpha=0.3)
    ax.set_title("Movement Path", fontsize=14, fontweight='bold')
    ax.set_aspect('equal')

    if len(path) > 1:
        # 경로 그리기 (초록색 점선)
        path_x = [pos[0] for pos in path]
        path_y = [pos[1] for pos in path]
        ax.plot(path_x, path_y, 'g--', linewidth=3, alpha=0.8,
               label=f'Path ({len(path)} points)')

        # 시작점 표시 (파란색 원)
        start_x, start_y = path[0]
        ax.plot(start_x, start_y, 'bo', markersize=10,
               label=f'Start ({start_x}, {start_y})')

        # 끝점 표시 (빨간색 원)
        end_x, end_y = path[-1]
        ax.plot(end_x, end_y, 'ro', markersize=10,
               label=f'Current ({end_x}, {end_y})')

        # 각 점에 번호 표시
        for i, (x, y) in enumerate(path):
            ax.text(x + 0.2, y + 0.2, str(i+1),
                   fontsize=10, fontweight='bold',
                   bbox=dict(boxstyle='round,pad=0.2',
                            facecolor='yellow', alpha=0.7))
    else:
        # 경로가 한 점뿐인 경우
        if path:
            x, y = path[0]
            ax.plot(x, y, 'go', markersize=15, label=f'Single point ({x}, {y})')
            ax.text(x + 0.2, y + 0.2, '1', fontsize=12, fontweight='bold')
        else:
            ax.text(0.5, 0.5, 'No path data', transform=ax.transAxes,
                   ha='center', va='center', fontsize=16,
                   bbox=dict(boxstyle='round', facecolor='lightgray'))

    # 경로 정보 출력
    if path:
        info_text = f"path_len: {len(path)}점\n"
        info_text += f"start: {path[0]}\n"
        info_text += f"now: {path[-1]}\n"
        if len(path) > 1:
            # 총 이동 거리 계산
            total_distance = 0
            for i in range(1, len(path)):
                x1, y1 = path[i-1]
                x2, y2 = path[i]
                distance = ((x2-x1)**2 + (y2-y1)**2)**0.5
                total_distance += distance
            info_text += f"total_distance: {total_distance:.2f}"

        ax.text(0.02, 0.98, info_text, transform=ax.transAxes,
               verticalalignment='top',
               bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))

    ax.legend()
    plt.tight_layout()
    plt.show()

    # 콘솔에도 정보 출력
    if path:
        print(f"경로: {path}")
        print(f"총 {len(path)}개 점을 지나감")

# 사용 예시
if __name__ == "__main__":
    # 기본 경로
    test_path = [(0, 0), (1, 1), (2, 2), (3, 3), (3, 4)]
    draw_path(test_path)

    # 지그재그 경로
    zigzag_path = [(0, 0), (1, 0), (1, 1), (2, 1), (2, 2), (3, 2)]
    draw_path(zigzag_path)

    # 한 점만 있는 경우
    single_path = [(2, 3)]
    draw_path(single_path)

    # 빈 경로
    draw_path([])
```

```python
import matplotlib.pyplot as plt

def draw_sensors(position, sensors):
    """센서 빔들만 그리는 함수"""
    plt.figure(figsize=(8, 6))
    ax = plt.gca()

    # 기본 설정
    ax.set_xlim(-1, 8)
    ax.set_ylim(-1, 12)
    ax.grid(True, alpha=0.3)
    ax.set_title("Sensor Beams", fontsize=14, fontweight='bold')
    ax.set_aspect('equal')

    x, y = position
    sensor_length = 3  # 최대 센서 길이

    # 자동차 위치 표시 (참고용)
    ax.plot(x, y, 'ko', markersize=10, label='Car Position')

    # 전방 센서 (파란색) - 위쪽으로
    front_reach = min(sensors['front'], sensor_length)
    ax.plot([x, x], [y, y + front_reach],
           'b-', linewidth=3, alpha=0.8, label=f'Front Sensor ({sensors["front"]:.1f})')

    # 좌측 센서 (초록색) - 왼쪽으로
    left_reach = min(sensors['left'], sensor_length)
    ax.plot([x, x - left_reach], [y, y],
           'g-', linewidth=3, alpha=0.8, label=f'Left Sensor ({sensors["left"]:.1f})')

    # 우측 센서 (빨간색) - 오른쪽으로
    right_reach = min(sensors['right'], sensor_length)
    ax.plot([x, x + right_reach], [y, y],
           'r-', linewidth=3, alpha=0.8, label=f'Right Sensor ({sensors["right"]:.1f})')

    # 센서 끝점에 점 표시
    ax.plot(x, y + front_reach, 'bo', markersize=8)  # 전방 끝점
    ax.plot(x - left_reach, y, 'go', markersize=8)   # 좌측 끝점
    ax.plot(x + right_reach, y, 'ro', markersize=8)  # 우측 끝점

    # 센서 값들 텍스트로 표시
    ax.text(x + 0.1, y + front_reach/2, f'{sensors["front"]:.1f}',
           color='blue', fontweight='bold')
    ax.text(x - left_reach/2 - 0.3, y + 0.1, f'{sensors["left"]:.1f}',
           color='green', fontweight='bold')
    ax.text(x + right_reach/2, y + 0.1, f'{sensors["right"]:.1f}',
           color='red', fontweight='bold')

    # 센서 정보 박스
    info_text = f"car_position: ({x}, {y})\n"
    info_text += f"front_sensor: {sensors['front']:.1f}\n"
    info_text += f"left_sensor: {sensors['left']:.1f}\n"
    info_text += f"right_sensor: {sensors['right']:.1f}\n"
    info_text += f"max_sensor_length: {sensor_length}"

    ax.text(0.02, 0.98, info_text, transform=ax.transAxes,
           verticalalignment='top',
           bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))

    ax.legend()
    plt.tight_layout()
    plt.show()

    print(f"센서 값들: {sensors}")
    print(f"자동차 위치: {position}")

# 사용 예시
if __name__ == "__main__":
    # 기본 센서 테스트
    test_sensors = {'front': 2.5, 'left': 1.2, 'right': 3.0}
    draw_sensors((3, 4), test_sensors)

    # 장애물이 가까운 경우
    close_sensors = {'front': 0.5, 'left': 0.8, 'right': 2.9}
    draw_sensors((2, 3), close_sensors)

    # 모든 방향이 막힌 경우
    blocked_sensors = {'front': 0.1, 'left': 0.2, 'right': 0.1}
    draw_sensors((1, 2), blocked_sensors)

    # 모든 방향이 열린 경우
    open_sensors = {'front': 3.0, 'left': 3.0, 'right': 3.0}
    draw_sensors((4, 2), open_sensors)
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle
import numpy as np

def create_visualization(step_data, obstacles):
    """
    자동차 시뮬레이션의 한 단계를 시각화하는 독립 함수

    Args:
        step_data (dict): 단계별 데이터
            - step: 단계 번호
            - position: (x, y) 현재 위치
            - action: 현재 행동
            - sensors: {'front': 값, 'left': 값, 'right': 값}
            - detected: 장애물 감지 여부
            - path: 지금까지의 경로 [(x1,y1), (x2,y2), ...]
            - goal_reached: 목표 도달 여부 (선택사항)
        obstacles (list): 장애물 위치 리스트 [(x1,y1), (x2,y2), ...]
    """
    plt.figure(figsize=(10, 8))
    ax = plt.gca()

    ax.set_xlim(-1, 8)
    ax.set_ylim(-1, 12)  # 목표 지점을 보기 위해 y축 확장
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('X Position', fontsize=12)
    ax.set_ylabel('Y Position', fontsize=12)
    ax.set_title(f"Step {step_data['step']} - Action: {step_data['action']}",
                fontsize=14, fontweight='bold')

    # Draw goal line (3 units above highest obstacle)
    max_obstacle_y = max(pos[1] for pos in obstacles) if obstacles else 0
    goal_y = max_obstacle_y + 3
    ax.axhline(y=goal_y, color='lime', linestyle='-', linewidth=3, alpha=0.7, label='Goal Line')
    ax.text(7, goal_y + 0.2, f'GOAL (y={goal_y})', ha='right', va='bottom',
            fontweight='bold', color='lime', fontsize=12)

    # Draw obstacles
    for i, (obs_x, obs_y) in enumerate(obstacles):
        rect = Rectangle((obs_x-0.4, obs_y-0.4), 0.8, 0.8,
                       color='red', alpha=0.7)
        ax.add_patch(rect)
        ax.text(obs_x, obs_y, str(i+1), ha='center', va='center',
               color='white', fontweight='bold')

    # Draw path
    if len(step_data['path']) > 1:
        path_x = [pos[0] for pos in step_data['path']]
        path_y = [pos[1] for pos in step_data['path']]
        ax.plot(path_x, path_y, 'g--', linewidth=2, alpha=0.7, label='Path')

    # Draw car
    x, y = step_data['position']

    # Car color based on action and goal status
    action = step_data['action']
    if step_data.get('goal_reached', False):
        car_color = 'gold'  # 목표 도달시 금색
    elif action == "stop":
        car_color = 'red'
    elif action == "turn_left":
        car_color = 'orange'
    elif action == "turn_right":
        car_color = 'purple'
    else:
        car_color = 'blue'

    car = Circle((x, y), 0.3, color=car_color)
    ax.add_patch(car)

    # Draw sensor beams
    sensors = step_data['sensors']
    sensor_length = 3
    ax.plot([x, x], [y, y + min(sensors['front'], sensor_length)],
           'b-', linewidth=3, alpha=0.8, label='Front Sensor')
    ax.plot([x, x - min(sensors['left'], sensor_length)], [y, y],
           'g-', linewidth=3, alpha=0.8, label='Left Sensor')
    ax.plot([x, x + min(sensors['right'], sensor_length)], [y, y],
           'r-', linewidth=3, alpha=0.8, label='Right Sensor')

    # Draw direction prediction (dotted line) - 간단한 예측 로직
    if action != "stop" and not step_data.get('goal_reached', False):
        # 간단한 다음 위치 예측
        if action == "forward" or action == "careful_forward":
            next_x, next_y = x, y + 1
        elif action == "turn_left":
            next_x, next_y = x - 1, y
        elif action == "turn_right":
            next_x, next_y = x + 1, y
        else:
            next_x, next_y = x, y

        ax.plot([x, next_x], [y, next_y], 'b:', linewidth=3, alpha=0.8, label='Next Direction')

    # Draw direction arrow
    if not step_data.get('goal_reached', False):
        if action == "forward" or action == "careful_forward":
            ax.arrow(x, y, 0, 0.5, head_width=0.1, head_length=0.1, fc='blue', ec='blue')
        elif action == "turn_left":
            ax.arrow(x, y, -0.5, 0, head_width=0.1, head_length=0.1, fc='orange', ec='orange')
        elif action == "turn_right":
            ax.arrow(x, y, 0.5, 0, head_width=0.1, head_length=0.1, fc='purple', ec='purple')

    # Display information
    info = f"Position: ({x}, {y})\n"
    info += f"Goal Y: {goal_y}\n"
    info += f"Sensors: F={sensors['front']:.1f}, L={sensors['left']:.1f}, R={sensors['right']:.1f}\n"
    info += f"Detected: {step_data['detected']}\n"
    info += f"Action: {action}"

    ax.text(0.02, 0.98, info, transform=ax.transAxes,
           verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat'))

    # Status message
    if step_data.get('goal_reached', False):
        status = "GOAL REACHED - SUCCESS!"
        status_color = 'gold'
    elif action == "stop":
        status = "STOPPED - All directions blocked!"
        status_color = 'red'
    elif step_data['detected']:
        status = "AVOIDING obstacles"
        status_color = 'orange'
    else:
        status = "Moving toward goal"
        status_color = 'lightgreen'

    ax.text(0.02, 0.02, status, transform=ax.transAxes, fontweight='bold',
           bbox=dict(boxstyle='round', facecolor=status_color, alpha=0.8))

    ax.legend()
    plt.tight_layout()
    plt.show()


# 사용 예시
if __name__ == "__main__":
    # 테스트 데이터 예시
    test_step_data = {
        'step': 5,
        'position': (3, 4),
        'action': 'turn_left',
        'sensors': {'front': 2.5, 'left': 1.2, 'right': 3.0},
        'detected': True,
        'path': [(0, 0), (1, 1), (2, 2), (3, 3), (3, 4)],
        'goal_reached': False
    }

    test_obstacles = [(2, 3), (4, 5), (1, 6), (5, 7)]

    # 시각화 실행
    create_visualization(test_step_data, test_obstacles)
```

```python
# 3-6번 라이브러리 추가됨

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle, Circle
import time

# 1.  센서 초기화 함수 # 키: "front", 값: 15.0 "left", 값: 15.0  "right", 값: 15.0
def init_sensors():
    """Initialize sensor default values"""
    sensors = {
        "front": 15.0,
        "left": 15.0,
        "right": 15.0}
    return sensors

# 2. 18-25 앞쪽 스캔
def scan_front(x, y, obstacles, sensors):
    """Scan front obstacles"""
    if (x, y+1) in obstacles:
        sensors["front"] = 1.0
    elif (x, y+2) in obstacles:
        sensors["front"] = 2.0
    elif (x, y+3) in obstacles:
        sensors["front"] = 3.0
# 3. 왼쪽 스캔
def scan_left(x, y, obstacles, sensors):
    """Scan left obstacles"""
    if (x-1, y) in obstacles:
        sensors["left"] = 1.0
    elif (x-2, y) in obstacles:
        sensors["left"] = 2.0
    elif (x-3, y) in obstacles:
        sensors["left"] = 3.0
# 4. 오른쪽 스캔
def scan_right(x, y, obstacles, sensors):
    """Scan right obstacles"""
    if (x+1, y) in obstacles:
        sensors["right"] = 1.0
    elif (x+2, y) in obstacles:
        sensors["right"] = 2.0
    elif (x+3, y) in obstacles:
        sensors["right"] = 3.0
# 5.센서 데이터 업데이트 함수. 현재 위치에서 주변 장애물을 스캔하여 센서 데이터를 업데이트
def update_sensor_data(position, obstacles):
    """Update sensor data from current position"""

    x, y = position["x"], position["y"]   #   현재 위치 좌표 추출

    sensors = init_sensors()   # 2. 센서 초기화 (기본값으로 설정),{"front": 15.0, "left": 15.0, "right": 15.0}

    scan_front(x, y, obstacles, sensors)     # 3.  전방 장애물 스캔
    scan_left(x, y, obstacles, sensors)    # 3. 좌측 장애물 스캔
    scan_right(x, y, obstacles, sensors)   # 3. 오른쪽 장애물 스캔

    return sensors    # 4. 업데이트된 센서 데이터 반환
# 6. 장애물 감지 함수
def detect_obstacles(sensors, safe_distance):
    """Detect obstacles based on sensor data"""
    # 감지된 장애물 방향을 저장할 빈 리스트 생성
    detected = []
      # 각 센서의 방향과 거리를 순회
    for direction, distance in sensors.items():
         # 안전거리보다 가까우면 위험 판단
        if distance < safe_distance:
            detected.append(direction)

    return detected     #  위험한 방향들의 리스트 반환
# 7. 행동 결정 함수
def decide_action(obstacles):
    """Decide action based on detected obstacles"""

    if len(obstacles) == 0:  #  장애물이 없는 경우
        action = "forward"   # 직진
    elif "front" in obstacles:       # 전방에 장애물이 있는 경우
        if "left" not in obstacles:
            action = "turn_left"
        elif "right" not in obstacles:
            action = "turn_right"
        else:
            action = "stop"
    #  전방은 안전하지만 좌/우측에 장애물이 있는 경우
    else:
        action = "careful_forward"   # 조심스럽게 직진
    return action

# 7-1. 추가된 부분 목표지점 확인하는 코드.
def check_goal_reached(current_position, obstacle_positions):
    """Check if car reached the goal (3 units above highest obstacle)"""
    if not obstacle_positions:
        return False

    max_obstacle_y = max(pos[1] for pos in obstacle_positions)
    goal_y = max_obstacle_y + 3

    return current_position["y"] >= goal_y

# 8. 자동 시뮬레이션 클래스 class SimpleCar:를 class AutoSimulationCar: 로 변경
class AutoSimulationCar:
    def __init__(self, name, x, y):
        """Initialize car for automatic simulation"""
        self.name = name
        self.position = {"x": x, "y": y}
        self.safe_distance = 10.0
        self.path_history = [(x, y)]  #추가된 부분
        self.step_count = 0  # 추가된 부분
        self.max_steps = 15  #추가된 부
        self.simulation_data = []  # 추가된 부분 모든 스텝 데이터 저장

    def move_forward(self):
        """Move forward"""
        self.position["y"] += 1
        self.path_history.append((self.position["x"], self.position["y"]))

    def turn_left(self):
        """Turn left"""
        self.position["x"] -= 1
        self.path_history.append((self.position["x"], self.position["y"]))

    def turn_right(self):
        """Turn right"""
        self.position["x"] += 1
        self.path_history.append((self.position["x"], self.position["y"]))
    # stop부분 삭제 액션이 "stop"이면 시뮬레이션 자체가 종료되므로, 별도의 stop() 메서드가 필요하지 않습니다.

    def execute_action(self, action):
        """Execute the decided action"""
        if action == "forward" or action == "careful_forward":
            self.move_forward()
        elif action == "turn_left":
            self.turn_left()
        elif action == "turn_right":
            self.turn_right()
        # stop does nothing
    # 추가된 부분  함수는 실제로 움직이지 않고 특정 행동을 했을 때의 다음 위치를 미리 계산하는 함수
    def get_next_position(self, action):
        """Get next position without actually moving"""
         # 1. 현재 위치 좌표 가져오기
        x, y = self.position["x"], self.position["y"]

           # 2. 각 행동별 다음 위치 계산 (실제 이동 X)
        if action == "forward" or action == "careful_forward":
            return (x, y + 1)   # 위로 1칸
        elif action == "turn_left":
            return (x - 1, y)  # 왼쪽으로 1칸
        elif action == "turn_right":
            return (x + 1, y)    # 오른쪽으로 1칸
        else:   # "stop" 등 기타 행동
            return (x, y)    # 현재 위치 그대로

    #  추가된 부분시각화하는 부분
    def create_visualization(self, step_data, obstacles):
        """Create visualization for one step"""
        plt.figure(figsize=(10, 8))
        ax = plt.gca()

        ax.set_xlim(-1, 8)
        ax.set_ylim(-1, 12)  # 목표 지점을 보기 위해 y축 확장
        ax.set_aspect('equal')
        ax.grid(True, alpha=0.3)
        ax.set_xlabel('X Position', fontsize=12)
        ax.set_ylabel('Y Position', fontsize=12)
        ax.set_title(f"Step {step_data['step']} - Action: {step_data['action']}",
                    fontsize=14, fontweight='bold')

        # Draw goal line (3 units above highest obstacle)
        max_obstacle_y = max(pos[1] for pos in obstacles)
        goal_y = max_obstacle_y + 3
        ax.axhline(y=goal_y, color='lime', linestyle='-', linewidth=3, alpha=0.7, label='Goal Line')
        ax.text(7, goal_y + 0.2, f'GOAL (y={goal_y})', ha='right', va='bottom',
                fontweight='bold', color='lime', fontsize=12)

        # Draw obstacles
        for i, (obs_x, obs_y) in enumerate(obstacles):
            rect = Rectangle((obs_x-0.4, obs_y-0.4), 0.8, 0.8,
                           color='red', alpha=0.7)
            ax.add_patch(rect)
            ax.text(obs_x, obs_y, str(i+1), ha='center', va='center',
                   color='white', fontweight='bold')

        # Draw path
        if len(step_data['path']) > 1:
            path_x = [pos[0] for pos in step_data['path']]
            path_y = [pos[1] for pos in step_data['path']]
            ax.plot(path_x, path_y, 'g--', linewidth=2, alpha=0.7, label='Path')

        # Draw car
        x, y = step_data['position']

        # Car color based on action and goal status
        action = step_data['action']
        if step_data.get('goal_reached', False):
            car_color = 'gold'  # 목표 도달시 금색
        elif action == "stop":
            car_color = 'red'
        elif action == "turn_left":
            car_color = 'orange'
        elif action == "turn_right":
            car_color = 'purple'
        else:
            car_color = 'blue'

        car = Circle((x, y), 0.3, color=car_color)
        ax.add_patch(car)

        # Draw sensor beams
        sensors = step_data['sensors']
        sensor_length = 3
        ax.plot([x, x], [y, y + min(sensors['front'], sensor_length)],
               'b-', linewidth=3, alpha=0.8, label='Front Sensor')
        ax.plot([x, x - min(sensors['left'], sensor_length)], [y, y],
               'g-', linewidth=3, alpha=0.8, label='Left Sensor')
        ax.plot([x, x + min(sensors['right'], sensor_length)], [y, y],
               'r-', linewidth=3, alpha=0.8, label='Right Sensor')

        # Draw direction prediction (dotted line)
        if action != "stop" and not step_data.get('goal_reached', False):
            next_x, next_y = self.get_next_position(action)
            ax.plot([x, next_x], [y, next_y], 'b:', linewidth=3, alpha=0.8, label='Next Direction')

        # Draw direction arrow
        if not step_data.get('goal_reached', False):
            if action == "forward" or action == "careful_forward":
                ax.arrow(x, y, 0, 0.5, head_width=0.1, head_length=0.1, fc='blue', ec='blue')
            elif action == "turn_left":
                ax.arrow(x, y, -0.5, 0, head_width=0.1, head_length=0.1, fc='orange', ec='orange')
            elif action == "turn_right":
                ax.arrow(x, y, 0.5, 0, head_width=0.1, head_length=0.1, fc='purple', ec='purple')

        # Display information
        info = f"Position: ({x}, {y})\n"
        info += f"Goal Y: {goal_y}\n"
        info += f"Sensors: F={sensors['front']:.1f}, L={sensors['left']:.1f}, R={sensors['right']:.1f}\n"
        info += f"Detected: {step_data['detected']}\n"
        info += f"Action: {action}"

        ax.text(0.02, 0.98, info, transform=ax.transAxes,
               verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat'))

        # Status message
        if step_data.get('goal_reached', False):
            status = "GOAL REACHED - SUCCESS!"
            status_color = 'gold'
        elif action == "stop":
            status = "STOPPED - All directions blocked!"
            status_color = 'red'
        elif step_data['detected']:
            status = "AVOIDING obstacles"
            status_color = 'orange'
        else:
            status = "Moving toward goal"
            status_color = 'lightgreen'

        ax.text(0.02, 0.02, status, transform=ax.transAxes, fontweight='bold',
               bbox=dict(boxstyle='round', facecolor=status_color, alpha=0.8))

        ax.legend()
        plt.tight_layout()
        plt.show()
#  자율주행 시뮬레이션
    def run_simulation(self, obstacles):
        """Run complete simulation and collect all data"""
        print("=== Autonomous Car Automatic Simulation ===")
        print(f"Obstacles at: {list(obstacles)}")
        print("Running simulation... Each step will be displayed automatically.")
        print()

        while self.step_count < self.max_steps:
            self.step_count += 1

            print(f"--- Step {self.step_count} ---")
            print(f"Current Position: ({self.position['x']}, {self.position['y']})")

            # Update sensor data
            sensors = update_sensor_data(self.position, obstacles)
            print(f"Sensors: {sensors}")

            # Detect obstacles
            detected_obstacles = detect_obstacles(sensors, self.safe_distance)
            print(f"Detected Obstacles: {detected_obstacles}")

            # Decide action
            action = decide_action(detected_obstacles)
            print(f"Action: {action}")

            # Store step data
            step_data = {
                'step': self.step_count,
                'position': (self.position['x'], self.position['y']),
                'sensors': sensors.copy(),
                'detected': detected_obstacles.copy(),
                'action': action,
                'path': self.path_history.copy()
            }

            # Show visualization
            self.create_visualization(step_data, obstacles)

            # If stopped, end simulation
            if action == "stop":
                print("STOPPED - All directions blocked!")
                break

            # Execute action
            self.execute_action(action)
            print(f"New Position: ({self.position['x']}, {self.position['y']})")
            print()

            # Small delay for better viewing
            time.sleep(1)

        print("Simulation Complete!")

    def run_all_at_once(self, obstacles):
        """Show all steps in sequence without interaction"""
        print("=== Showing All Steps Automatically ===")
        print(f"Goal: Reach y-coordinate {max(pos[1] for pos in obstacles) + 3}")
        print("Each step will be displayed for 2 seconds")
        print()

        # Run simulation and collect data first
        steps_data = []
        temp_position = self.position.copy()
        temp_path = self.path_history.copy()

        for step in range(self.max_steps):
            step_num = step + 1

            # Check if goal is reached
            goal_reached = check_goal_reached(temp_position, obstacles)
            if goal_reached:
                print(f"GOAL REACHED at position ({temp_position['x']}, {temp_position['y']})!")
                # Add final step showing goal reached
                sensors = update_sensor_data(temp_position, obstacles)
                step_data = {
                    'step': step_num,
                    'position': (temp_position['x'], temp_position['y']),
                    'sensors': sensors.copy(),
                    'detected': [],
                    'action': "goal_reached",
                    'path': temp_path.copy(),
                    'goal_reached': True
                }
                steps_data.append(step_data)
                break

            # Update sensor data
            sensors = update_sensor_data(temp_position, obstacles)

            # Detect obstacles
            detected_obstacles = detect_obstacles(sensors, self.safe_distance)

            # Decide action
            action = decide_action(detected_obstacles)

            # Store step data
            step_data = {
                'step': step_num,
                'position': (temp_position['x'], temp_position['y']),
                'sensors': sensors.copy(),
                'detected': detected_obstacles.copy(),
                'action': action,
                'path': temp_path.copy(),
                'goal_reached': False
            }
            steps_data.append(step_data)

            # If stopped, end
            if action == "stop":
                print(f"STOPPED at position ({temp_position['x']}, {temp_position['y']}) - All directions blocked!")
                break

            # Execute action on temp position
            if action == "forward" or action == "careful_forward":
                temp_position["y"] += 1
                temp_path.append((temp_position["x"], temp_position["y"]))
            elif action == "turn_left":
                temp_position["x"] -= 1
                temp_path.append((temp_position["x"], temp_position["y"]))
            elif action == "turn_right":
                temp_position["x"] += 1
                temp_path.append((temp_position["x"], temp_position["y"]))

        # Now show all steps
        for i, step_data in enumerate(steps_data):
            if step_data.get('goal_reached', False):
                print(f"Showing Final Step - GOAL REACHED!")
            else:
                print(f"Showing Step {step_data['step']}")
            self.create_visualization(step_data, obstacles)
            if i < len(steps_data) - 1:
                time.sleep(2)  # 2초 대기

        if steps_data and steps_data[-1].get('goal_reached', False):
            print("SUCCESS! Car reached the goal!")
        else:
            print("Simulation ended.")
        print("All steps completed!")

# 실행
if __name__ == "__main__":
    # 자동차 생성
    car = AutoSimulationCar("TestCar", 3, 0)

    # 장애물 설정
    obstacles = {(3, 3), (3, 4), (2, 5), (4, 6)} # 더 멀리 배치
    # 3-6번 라이브러리 추가됨


    car.run_all_at_once(obstacles)
```
