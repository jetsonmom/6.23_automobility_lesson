NumPy: CPU만, 과학계산 전용

TensorFlow: GPU 지원, 딥러닝 전용, 자동 미분 기능

TensorFlow 텐서 기본 사용법
 TensorFlow는 NumPy를 참고해서 설계함.
## 주요 내용

**1. 텐서 생성하기**
- 스칼라(0차원): 숫자 하나
- 벡터(1차원): 일렬로 나열된 숫자들  
- 행렬(2차원): 표 형태
- 3차원 텐서: 여러 행렬을 쌓은 형태

**2. 텐서 정보 확인**
- shape: 텐서의 크기와 형태
- dtype: 데이터 타입 (정수, 실수 등)
- rank: 몇 차원인지
- size: 전체 원소 개수

**3. 다양한 생성 방법**
- `tf.zeros()`: 0으로 채우기
- `tf.ones()`: 1로 채우기
- `tf.fill()`: 특정 값으로 채우기
- `tf.range()`: 연속된 숫자
- `tf.random.normal()`: 랜덤 값

**4. 텐서 접근하기**
- 인덱싱으로 특정 원소나 영역 선택
- 파이썬 리스트처럼 `[행, 열]` 방식 사용

**5. 데이터 타입**
- int32, float32, bool, string 등


# NumPy ↔ TensorFlow 비교표

## 기본 생성 함수
| 기능 | NumPy | TensorFlow |
|------|-------|------------|
| 0으로 채우기 | `np.zeros((2, 3))` | `tf.zeros([2, 3])` |
| 1로 채우기 | `np.ones((2, 3))` | `tf.ones([2, 3])` |
| 배열/텐서 생성 | `np.array([1,2,3])` | `tf.constant([1,2,3])` |
| 특정값으로 채우기 | `np.full((2, 3), 7)` | `tf.fill([2, 3], 7)` |
| 연속 숫자 | `np.arange(1, 11)` | `tf.range(1, 11)` |
| 등간격 숫자 | `np.linspace(0, 10, 5)` | `tf.linspace(0.0, 10.0, 5)` |

## 랜덤 함수
| 기능 | NumPy | TensorFlow |
|------|-------|------------|
| 정규분포 | `np.random.normal(0, 1, (2,3))` | `tf.random.normal([2,3], 0, 1)` |
| 균등분포 | `np.random.uniform(0, 10, (2,3))` | `tf.random.uniform([2,3], 0, 10)` |
| 시드 설정 | `np.random.seed(42)` | `tf.random.set_seed(42)` |

## 정보 확인
| 기능 | NumPy | TensorFlow |
|------|-------|------------|
| 형태 | `arr.shape` | `tensor.shape` |
| 차원 수 | `arr.ndim` | `tf.rank(tensor)` |
| 원소 개수 | `arr.size` | `tf.size(tensor)` |
| 데이터 타입 | `arr.dtype` | `tensor.dtype` |

## 기본 연산
| 기능 | NumPy | TensorFlow |
|------|-------|------------|
| 덧셈 | `a + b` | `tf.add(a, b)` 또는 `a + b` |
| 곱셈 | `a * b` | `tf.multiply(a, b)` 또는 `a * b` |
| 행렬곱 | `np.matmul(a, b)` | `tf.matmul(a, b)` |
| 전치 | `a.T` | `tf.transpose(a)` |

## 형태 변환
| 기능 | NumPy | TensorFlow |
|------|-------|------------|
| 형태 변경 | `arr.reshape(2, 3)` | `tf.reshape(tensor, [2, 3])` |
| 차원 추가 | `np.expand_dims(arr, 0)` | `tf.expand_dims(tensor, 0)` |
| 차원 제거 | `np.squeeze(arr)` | `tf.squeeze(tensor)` |

## 인덱싱
| 기능 | NumPy | TensorFlow |
|------|-------|------------|
| 단일 원소 | `arr[0, 1]` | `tensor[0, 1]` |
| 행 선택 | `arr[0, :]` | `tensor[0, :]` |
| 열 선택 | `arr[:, 1]` | `tensor[:, 1]` |
| 슬라이싱 | `arr[1:3, 0:2]` | `tensor[1:3, 0:2]` |

## 상호 변환
| 변환 방향 | 코드 |
|-----------|------|
| NumPy → TensorFlow | `tf.constant(numpy_array)` |
| TensorFlow → NumPy | `tensor.numpy()` |

## 주요 차이점
| 항목 | NumPy | TensorFlow |
|------|-------|------------|
| 실행 환경 | CPU만 | CPU + GPU |
| 주 사용 목적 | 과학 계산 | 딥러닝 |
| 자동 미분 | ❌ | ✅ |
| 즉시 실행 | ✅ | ✅ (2.0+) |
| 모델 저장 | ❌ | ✅ |
| 분산 처리 | ❌ | ✅ |

**💡 팁**: NumPy를 알고 있다면 TensorFlow는 "GPU 지원 + 딥러닝 기능이 추가된 NumPy"라고 생각하면 됩니다!

```python
# TensorFlow 텐서 기본 사용법
import tensorflow as tf
import numpy as np

print("=== TensorFlow 텐서 기본 사용법 ===")
# 1. 다양한 차원의 텐서 생성
print("\n1. 텐서 생성하기")
print("-" * 30)
```

```python
# 0차원 텐서 (스칼라) - 숫자 하나
scalar = tf.constant(5)
print(f"스칼라 (0차원): {scalar}")
print(f"스칼라 shape: {scalar.shape}")
print(f"스칼라 값: {scalar.numpy()}")
```

```python
# 1차원 텐서 (벡터) - 일렬로 나열된 숫자들
vector = tf.constant([1, 2, 3, 4, 5])
print(f"\n벡터 (1차원): {vector}")
print(f"벡터 shape: {vector.shape}")
print(f"벡터 길이: {len(vector)}")
```

```python
# 2차원 텐서 (행렬) - 표 형태의 숫자들
matrix = tf.constant([[1, 2, 3],
                     [4, 5, 6]])
print(f"\n행렬 (2차원):\n{matrix}")
print(f"행렬 shape: {matrix.shape}")
print(f"행렬 크기: {matrix.shape[0]}행 x {matrix.shape[1]}열")
```

```python
# 3차원 텐서 - 여러 개의 행렬을 쌓은 형태
tensor_3d = tf.constant([[[1, 2], [3, 4]],
                        [[5, 6], [7, 8]]])
print(f"\n3차원 텐서:\n{tensor_3d}")
print(f"3차원 텐서 shape: {tensor_3d.shape}")
print(f"3차원 텐서 크기: {tensor_3d.shape[0]} x {tensor_3d.shape[1]} x {tensor_3d.shape[2]}")
```

```python
# 2. 텐서 정보 확인하기
print("\n\n2. 텐서 정보 확인하기")
print("-" * 30)

sample_tensor = tf.constant([[1.5, 2.7, 3.1],
                            [4.2, 5.8, 6.3]])

print(f"텐서:\n{sample_tensor}")
print(f"Shape (형태): {sample_tensor.shape}")
print(f"Dtype (데이터 타입): {sample_tensor.dtype}")
print(f"Rank (차원 수): {tf.rank(sample_tensor)}")
print(f"Size (전체 원소 개수): {tf.size(sample_tensor)}")
```

```python
# 3. 다양한 방법으로 텐서 만들기
print("\n\n3. 다양한 방법으로 텐서 만들기")
print("-" * 30)

# 모든 값이 0인 텐서
zeros = tf.zeros([2, 3])
print(f"0으로 채운 텐서 (2x3):\n{zeros}")
```

```python
# 모든 값이 1인 텐서
ones = tf.ones([3, 2])
print(f"\n1로 채운 텐서 (3x2):\n{ones}")
```

```python
# 특정 값으로 채운 텐서
filled = tf.fill([2, 4], 7)
print(f"\n7로 채운 텐서 (2x4):\n{filled}")
```

```python
# 순서대로 증가하는 텐서
range_tensor = tf.range(1, 11)
print(f"\n1부터 10까지 텐서: {range_tensor}")
```

```python
# 정규분포 랜덤 텐서
random_normal = tf.random.normal([2, 3], mean=0.0, stddev=1.0)
print(f"\n정규분포 랜덤 텐서 (2x3):\n{random_normal}")
```

```python

# 균등분포 랜덤 텐서
random_uniform = tf.random.uniform([2, 3], minval=0, maxval=10)
print(f"\n균등분포 랜덤 텐서 (2x3):\n{random_uniform}")
```

```python
# 4. 텐서 접근하기 (인덱싱)
print("\n\n4. 텐서 접근하기")
print("-" * 30)

sample_matrix = tf.constant([[10, 20, 30],
                            [40, 50, 60],
                            [70, 80, 90]])

print(f"원본 행렬:\n{sample_matrix}")
print(f"첫 번째 행: {sample_matrix[0]}")
print(f"두 번째 열: {sample_matrix[:, 1]}")
print(f"특정 원소 (1행 2열): {sample_matrix[1, 2]}")
print(f"일부 영역 (처음 2행, 마지막 2열):\n{sample_matrix[:2, 1:]}")
```

```python
# 5. 실제 데이터 타입 예제
print("\n\n5. 실제 데이터 타입 예제")
print("-" * 30)
```

```python
# 정수 타입
int_tensor = tf.constant([1, 2, 3], dtype=tf.int32)
print(f"정수 텐서: {int_tensor}, 타입: {int_tensor.dtype}")
```

```python

# 실수 타입
float_tensor = tf.constant([1.0, 2.0, 3.0], dtype=tf.float32)
print(f"실수 텐서: {float_tensor}, 타입: {float_tensor.dtype}")
```

```python
# 불린 타입
bool_tensor = tf.constant([True, False, True], dtype=tf.bool)
print(f"불린 텐서: {bool_tensor}, 타입: {bool_tensor.dtype}")
```

```python
# 문자열 타입
string_tensor = tf.constant(["안녕", "하세요", "텐서"], dtype=tf.string)
print(f"문자열 텐서: {string_tensor}, 타입: {string_tensor.dtype}")

print("\n" + "="*50)
print("이제 이 코드를 코랩에서 실행해보세요!")
print("각 섹션별로 천천히 실행하면서 결과를 확인해보시면 됩니다.")
```
