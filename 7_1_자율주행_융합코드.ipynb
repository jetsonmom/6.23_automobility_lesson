```python
복# 1. 센서 초기화 함수
def init_sensors():
    """센서 기본값 설정"""
    sensors = {
        "앞": 15.0,      # 기본값
        "왼쪽": 15.0,
        "오른쪽": 15.0
    }
    print("1-09 - 센서 초기화 완료", sensors)
    return sensors잡한 코드 이 코드는 센서 기반 자율주행 자동차 시뮬레이션 코드
```

```python
# 2.앞쪽 스캔 함수
def scan_front(x, y, obstacles, sensors):
    """앞쪽 장애물 스캔"""
    print("2-04 - 앞쪽 스캔 시작")
    if (x, y+1) in obstacles:
        print("2-06", x, y, "앞쪽 1칸 장애물 발견")
        sensors["앞"] = 1.0
    elif (x, y+2) in obstacles:
        sensors["앞"] = 2.0
        print("2-10", x, y, "앞쪽 2칸 장애물 발견")
    elif (x, y+3) in obstacles:
        sensors["앞"] = 3.0
        print("2-13", x, y, "앞쪽 3칸 장애물 발견")
    else:
        print("2-15", x, y, "앞쪽 장애물 없음")
    print("2-16 - 앞쪽 스캔 완료")
```

3. 왼쪽 스캔 함수 https://claude.ai/public/artifacts/0bd88961-bc8c-43b4-b40b-aa0a27fe96df

```python
# 3. 왼쪽 스캔 함수
def scan_left(x, y, obstacles, sensors):
    """왼쪽 장애물 스캔"""
    print("3-4 - 왼쪽 스캔 시작")
    if (x-1, y) in obstacles:
        sensors["왼쪽"] = 1.0
        print("3-7", x, y, "왼쪽 1칸 장애물 발견")
    elif (x-2, y) in obstacles:
        sensors["왼쪽"] = 2.0
        print("3-10", x, y, "왼쪽 2칸 장애물 발견")
    elif (x-3, y) in obstacles:
        sensors["왼쪽"] = 3.0
        print("3-13", x, y, "왼쪽 3칸 장애물 발견")
    else:
        print("3-15", x, y, "왼쪽 장애물 없음")
    print("44 - 왼쪽 스캔 완료")
```

```python
def scan_right(x, y, obstacles, sensors):
    """오른쪽 장애물 스캔"""
    print("4-4 - 오른쪽 스캔 시작")
    if (x+1, y) in obstacles:
        sensors["오른쪽"] = 1.0
        print("4-7", x, y, "오른쪽 1칸 장애물 발견")
    elif (x+2, y) in obstacles:
        sensors["오른쪽"] = 2.0
        print("4-10", x, y, "오른쪽 2칸 장애물 발견")
    elif (x+3, y) in obstacles:
        sensors["오른쪽"] = 3.0
        print("4-13", x, y, "오른쪽 3칸 장애물 발견")
    else:
        print("4-15", x, y, "오른쪽 장애물 없음")
    print("4-16 - 오른쪽 스캔 완료")
```

 5. 센서 데이터 업데이트 함수 (디버그 print 포함) flowchart  https://claude.ai/public/artifacts/0071fea5-0417-4ac3-b753-5e47ea108c2a

```python
# 5. 센서 데이터 업데이트 함수 (디버그 print 포함)
def update_sensor_data(position, obstacles):
    """현재 위치에서 센서 데이터 업데이트"""
    print("5-4 - update_sensor_data 함수 시작")
    x, y = position["x"], position["y"]
    print("5-6- 위치 추출 완료")
    print("5-7", x, y)

    # 센서 초기화
    sensors = init_sensors()

    # 각 방향별 스캔 실행
    scan_front(x, y, obstacles, sensors)
    scan_left(x, y, obstacles, sensors)
    scan_right(x, y, obstacles, sensors)

    print("5-17 - 센서 업데이트 완료", sensors)
    return sensors
```

```python
# 6.
def detect_obstacles(sensors, safe_distance):
    """장애물을 감지하는 함수 (원본 그대로)"""
    print("6-4 - detect_obstacles 함수 시작")
    print("6-5", sensors, safe_distance)
    obstacles = []

    for direction, distance in sensors.items():
        print("6-9", direction, distance, "체크 중")
        if distance < safe_distance:
            obstacles.append(direction)
            print("6-12", direction, "위험! 장애물 목록에 추가")
        else:
            print("6-14", direction, "안전")

    print("6-16 - 장애물 감지 완료", obstacles)
    return obstacles
```

```python
# 7
def decide_action(obstacles):
    """장애물 정보를 바탕으로 행동을 결정하는 함수 (원본 그대로)"""
    print("7-4 - decide_action 함수 시작")
    print("7-5", obstacles, "입력된 장애물 목록")

    if len(obstacles) == 0:
        action = "직진"
        print("7-9", "장애물 없음 -> 직진")
    elif "앞" in obstacles:
        print("7-11", "앞에 장애물 있음, 좌우 확인")
        if "왼쪽" not in obstacles:
            action = "좌회전"
            print("7-14", "왼쪽 안전 -> 좌회전")
        elif "오른쪽" not in obstacles:
            action = "우회전"
            print("7-17", "오른쪽 안전 -> 우회전")
        else:
            action = "정지"
            print("7-20", "모든 방향 위험 -> 정지")
    else:
        action = "주의해서 직진"
        print("7-23", "앞은 안전, 옆만 위험 -> 주의해서 직진")

    print("7-25 - 행동 결정 완료", action)
    return action
```

```python
# 8. 간단한 자율주행 자동차 클래스 (디버그 포함)
class SimpleCar:
    def __init__(self, name, x, y):
        """자동차 생성"""
        print("8-5 - SimpleCar __init__ 시작")
        self.name = name
        self.position = {"x": x, "y": y}
        self.safe_distance = 10.0
        print("8-8", name, x, y, "자동차 생성 완료")

    def move_forward(self):
        """전진"""
        print("8-13 - move_forward 시작")
        self.position["y"] += 1
        print("8-15", self.position["x"], self.position["y"], "전진 완료")
        print(f"🚗 {self.name}이(가) 전진합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def turn_left(self):
        """좌회전"""
        print("8-20 - turn_left 시작")
        self.position["x"] -= 1
        print("8-22", self.position["x"], self.position["y"], "좌회전 완료")
        print(f"↖️ {self.name}이(가) 좌회전합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def turn_right(self):
        """우회전"""
        print("8-27 - turn_right 시작")
        self.position["x"] += 1
        print("8-29", self.position["x"], self.position["y"], "우회전 완료")
        print(f"↗️ {self.name}이(가) 우회전합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def stop(self):
        """정지"""
        print("8-34 - stop 시작")
        print("8-35", self.position["x"], self.position["y"], "정지")
        print(f"🛑 {self.name}이(가) 정지합니다. 위치: ({self.position['x']}, {self.position['y']})")

    def execute_action(self, action):
        """행동 실행"""
        print("8-40 - execute_action 시작")
        print("8-41", action, "실행할 행동")

        if action == "직진" or action == "주의해서 직진":
            print("8-44", "직진 관련 행동 -> move_forward 호출")
            self.move_forward()
        elif action == "좌회전":
            print("8-47", "좌회전 -> turn_left 호출")
            self.turn_left()
        elif action == "우회전":
            print("8-50", "우회전 -> turn_right 호출")
            self.turn_right()
        elif action == "정지":
            print("8-53", "정지 -> stop 호출")
            self.stop()

        print("8-56 - execute_action 완료")

    def run_simulation(self, obstacles):
        """자율주행 시뮬레이션"""
        print("8-60 - run_simulation 시작")
        steps = 0
        max_steps = 10

        print("🚗 자율주행 장애물 감지 시스템")
        print("-" * 30)
        print(f"🚧 장애물 위치: {list(obstacles)}")
        print("8-67", obstacles, "시뮬레이션 시작")

        while steps < max_steps:
            print(f"\n--- 스텝 {steps + 1} ---")
            print("8-71", steps + 1, "스텝 시작")
            print(f"📍 현재 위치: ({self.position['x']}, {self.position['y']})")
            print("8-73", self.position["x"], self.position["y"], "현재 위치")

            # 센서 데이터 업데이트
            print("8-76 - 센서 업데이트 호출")
            sensors = update_sensor_data(self.position, obstacles)
            print(f"📡 센서 데이터: {sensors}")
            print("8-79", sensors, "센서 데이터 받음")

            # 장애물 감지 (원본 함수 그대로 사용)
            print("8-82 - 장애물 감지 호출")
            detected_obstacles = detect_obstacles(sensors, self.safe_distance)
            print(f"🚨 감지된 장애물: {detected_obstacles}")
            print("8-85", detected_obstacles, "감지된 장애물")

            # 행동 결정 (원본 함수 그대로 사용)
            print("8-88 - 행동 결정 호출")
            action = decide_action(detected_obstacles)
            print(f"🎯 결정된 행동: {action}")
            print("8-91", action, "결정된 행동")

            # 행동 실행
            print("8-94 - 행동 실행 호출")
            self.execute_action(action)

            # 정지했으면 종료
            if action == "정지":
                print("8-99", "정지 조건 만족, 시뮬레이션 종료")
                print("⚠️ 모든 방향 막혀서 시뮬레이션 종료!")
                break

            steps += 1
            print("8-104", steps, "스텝 완료")

        print(f"\n🏁 시뮬레이션 완료!")
        print("8-107 - run_simulation 완료")

```

```python
# 9. 시뮬레이션 실행
if __name__ == "__main__":
    print("9-3 - 메인 실행 시작")

    # 자동차 생성
    print("9-6 - 자동차 생성")
    car = SimpleCar("현대차", 3, 0)

    # 장애물 설정
    print("9-10 - 장애물 설정")
    obstacles = {(3, 2), (3, 3), (2, 4), (4, 5)}
    print("9-12", obstacles, "장애물 설정 완료")

    # 시뮬레이션 실행
    print("9-15 - 시뮬레이션 호출")
    car.run_simulation(obstacles)

    print("9-18 - 프로그램 종료")
```
